```python
import cv2
import numpy as np

def sketch(image):
    # Convert the image to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    # Apply Gaussian Blur
    blur_gray = cv2.GaussianBlur(gray, (5, 5), 900)
    # Detect edges using Canny
    edges = cv2.Canny(blur_gray, 45, 90)
    # Apply binary inverse threshold
    ret, thre = cv2.threshold(edges, 70, 255, cv2.THRESH_BINARY_INV)
    return thre

# Initialize webcam
cam = cv2.VideoCapture(0)

print("Press 'Enter' to save the sketch.")
print("Press 'Esc' to exit the application.")

while True:
    ret, frame = cam.read()
    if not ret:
        print("Failed to capture video. Exiting...")
        break

    # Display the live sketch
    cv2.imshow('Live Sketch', sketch(frame))

    # Wait for user input
    key = cv2.waitKey(1)
    if key == 27:  # Esc key
        print("Exiting...")
        break
    elif key == 13:  # Enter key
        cv2.imwrite('sketch.jpg', sketch(frame))
        print("Image Saved!!!")

# Release resources
cam.release()
cv2.destroyAllWindows()
```
